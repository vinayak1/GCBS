```python
# (4) CosQA Contrastive pretraining
# See here for CSN and AdvTest instructions <May need to change data_file paths>:
# https://github.com/microsoft/CodeBERT/tree/master/UniXcoder/downstream-tasks/code-search#fine-tune-setting

%cd /content/drive/MyDrive/unixcoder_exps/CodeBERT/UniXcoder/downstream-tasks/code-search
# Training
!python run.py \
    --output_dir saved_models/cosqa \
    --model_name_or_path microsoft/unixcoder-base  \
    --do_train \
    --train_data_file ../cosqa/cosqa-retrieval-train-19604.json \
    --eval_data_file ../cosqa/cosqa-retrieval-dev-500.json \
    --codebase_file ../cosqa/code_idx_map.txt \
    --num_train_epochs 10 \
    --gcbs_quantile 0.999 \
    --code_length 256 \
    --nl_length 128 \
    --train_batch_size 64 \
    --eval_batch_size 64 \
    --learning_rate 2e-5 \
    --seed 123456

# Hyperparameters to tune for all 3 datasets (cmd line arguments)
# (1) Learning rate: --learning_rate [range: (5e-6, 1e-4)]
# (2) Batch Size: --train_batch_size [32, 64, 128, 256, 512]
# (3) GCBS Quantile: --gcbs_quantile [0.995, 0.999, 0.9995, 0.9999]
# (4) Number of Epochs: --num_train_epochs [1, 2, 3, 5, 7, 10]
```
